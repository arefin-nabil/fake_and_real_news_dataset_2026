```python
import pandas as pd
import numpy as np
```

```python
df_real = pd.read_csv("real_raw.csv", encoding='utf-8-sig')
df_fake = pd.read_csv("fake_raw.csv", encoding='utf-8-sig')
```

```python
print("Real : ", df_real.shape)
print("Fake : ", df_fake.shape)
```

```python
print("Real : ", df_real.columns)
print("Fake : ", df_fake.columns)
```

```python
# real=0, fake=1
df_real['label'] = 0
df_fake['label'] = 1
```

```python
df = pd.concat([df_real, df_fake], ignore_index=True)
```

```python
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
print("Real and Fake : ", df.shape)
```

```python
df['label'].value_counts()
```

```python
df.isnull().sum()
```

```python
df.to_csv("full_dataset.csv", index=False, encoding='utf-8-sig')
```
